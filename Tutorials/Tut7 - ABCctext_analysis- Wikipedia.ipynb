# Basics of Text Manipulation

- Most text manipulation in Python can be achieved through the built-in methods of the `str` class plus regular expressions.
- Regular expressions using the `re` module have already been covered, so we will use it mostly in examples.
- Note: to remove strings, you replace them with `''`
    - `re.sub(pattern, replacement, text)`: for substituting all instances of `pattern` in `text` with `replacement`.
    - `str.replace(from, to)`: like the above, but for specific text `from`. Cleaner than `sub` if regex is not needed.
- [Documentation](https://docs.python.org/3/library/stdtypes.html#text-sequence-type-str])

# Intermediate Processing

- Most text cleaning can be done using the above, although you might want to make non-obvious changes to text.
- Still, there are non-obvious operations that are useful when doing text analysis.
- The Natural Languate ToolKit (NLTK) library in Python provides several tools for this kind of processing.
- [Documentation](https://www.nltk.org/)

## Stemming, POS Tagging and Tokenizing
- Tokenizing
    - Splitting tentences into words is more complicated than it seems.
    - What makes up a sentence, or even a single word, isn't so clear, as punctuation can vary.
    - We might like to split "I'll" into two parts, as it really means two words.
    - Numbers are also meant to be a single part, even though a fraction might have a dot in between.
- Stemming
    - Replace words with their "core" parts.
    - Multiple english words can have the same semantic meaning, but have a few letters that
    - This way, "books" and "book" can be replaced by the same word.
    - Useful for when you want to reduce the number of words you consider in an analysis.
- Part of Speach (POS) Tagging
    - Replace words by their "linguistic groups", such as noun phrases.
    - Useful for replacing words with their group in instances when the group does not matter.
        - Like when a name doesn't matter as much as the fact that it is a name.
     
## Notes
- How you process text is always dependent on your objective.
    - Text to be analyzed statically should be treated differently than text to be used in an ML model.
    - Treatment of puntuation, tokens and such should be crafted for your specific objectives. 
- While the above are some of the most used processing tools, they are just the tip of the iceberg.
- We will likely cover more advanced features later, but don't hold back on exploring the library yourself.
- As mentioned, these are "non-obvious" so there are actually several different methods you can choose from to acomplish the same task.
    - I will only use the "standard" methods here.
    - If you want to perform more serious text analysis, make sure to understand the other options.

```python
import wikipediaapi

def get_wikipedia_content(page_name):
    # Creating Wikipedia object with a specific user-agent
    wiki = wikipediaapi.Wikipedia(
        language='en',
        extract_format=wikipediaapi.ExtractFormat.WIKI,
        user_agent="MyPythonApp/1.0 (contact: email@example.com)"
    )
    page = wiki.page(page_name)
    if page.exists():
        return page.text
    else:
        return "Page does not exist."

# Get content from the Wikipedia pages
decision_tree_text = get_wikipedia_content("Decision tree").lower()
random_forest_text = get_wikipedia_content("Random forest").lower()
boosting_trees_text = get_wikipedia_content("Boosting (machine learning)").lower()

print("Decision Tree Content Length:", len(decision_tree_text))
print("Random Forest Content Length:", len(random_forest_text))
print("Boosting Trees Content Length:", len(boosting_trees_text))
```

```python
def count_word_occurrences(text, word):
    return text.lower().count(word.lower())

tree_count_decision = count_word_occurrences(decision_tree_text, "tree")
tree_count_forest = count_word_occurrences(random_forest_text, "tree")
tree_count_boosting = count_word_occurrences(boosting_trees_text, "tree")

print("Number of times 'tree' appears in Decision Tree:", tree_count_decision)
print("Number of times 'tree' appears in Random Forest:", tree_count_forest)
print("Number of times 'tree' appears in Boosting Trees:", tree_count_boosting)
```

```python
def remove_words(name, text, words_to_remove):
    counts = {}
    for word in words_to_remove:
        before = text.count(f" {word} ") + text.count(f"\n{word} ")
        counts[word] = before
        text = text.replace(f" {word} ", " ").replace(f"\n{word} ", "\n ")
    print(f"{name}: " + ", ".join(f"{w}:{c}" for w, c in counts.items() if c))
    return text

words_to_remove = ["we", "our", "they", "the", "a", "their", "what", "which", "that"]

decision_tree_text = remove_words("DecisionTree", decision_tree_text, words_to_remove)
random_forest_text = remove_words("RandomForest", random_forest_text, words_to_remove)
boosting_trees_text = remove_words("BoostingTrees", boosting_trees_text, words_to_remove)
```

```python
import string

def remove_punctuation(text):
    removed = [ch for ch in text if ch in string.punctuation]
    print(f"Removed {len(removed)} punctuations: {''.join(sorted(set(removed)))}")
    return text.translate(str.maketrans("", "", string.punctuation))

for name in ["decision_tree_text", "random_forest_text", "boosting_trees_text"]:
    locals()[name] = remove_punctuation(locals()[name])

```

```python
def remove_numbers(name, text):
    digits = [ch for ch in text if ch.isdigit()]
    print(f"{name}: removed {len(digits)} digits ({''.join(sorted(set(digits)))})")
    return ''.join(ch for ch in text if not ch.isdigit())

decision_tree_text = remove_numbers("DecisionTree", decision_tree_text)
random_forest_text = remove_numbers("RandomForest", random_forest_text)
boosting_trees_text = remove_numbers("BoostingTrees", boosting_trees_text)
```

```python
def split_text_into_words(text):
    return set(text.lower().split())

# Splitting each text into sets of words
decision_tree_words = split_text_into_words(decision_tree_text)
random_forest_words = split_text_into_words(random_forest_text)
boosting_trees_words = split_text_into_words(boosting_trees_text)
decision_tree_text[:1000],decision_tree_words
```

```python
from collections import Counter

def analyze_text(words):
    word_list = words.split()
    word_count = len(word_list)
    word_freq = Counter(word_list)
    most_common_word, highest_freq = word_freq.most_common(1)[0]
    return word_count, word_freq, most_common_word, highest_freq

# Analyzing each text
decision_word_count, decision_word_freq, decision_most_common, decision_highest_freq = analyze_text(decision_tree_text)
forest_word_count, forest_word_freq, forest_most_common, forest_highest_freq = analyze_text(random_forest_text)
boosting_word_count, boosting_word_freq, boosting_most_common, boosting_highest_freq = analyze_text(boosting_trees_text)

print("Decision Tree Analysis:")
print("Total number of words:", decision_word_count)
print("Most common word:", decision_most_common, "with frequency:", decision_highest_freq)

print("\nRandom Forest Analysis:")
print("Total number of words:", forest_word_count)
print("Most common word:", forest_most_common, "with frequency:", forest_highest_freq)

print("\nBoosting Trees Analysis:")
print("Total number of words:", boosting_word_count)
print("Most common word:", boosting_most_common, "with frequency:", boosting_highest_freq)
```

```python
from collections import Counter

def analyze_text(name, text):
    words = text.split()
    freq = Counter(words)
    total = len(words)
    unique = len(freq)
    most_common = freq.most_common(5)
    avg_freq = total / unique if unique else 0

    print(f"\n{name} Analysis:")
    print(f"Total words: {total:,}")
    print(f"Unique words: {unique:,}")
    print(f"Average frequency per word: {avg_freq:.2f}")
    print("Top 5 words:", ", ".join(f"{w}({c})" for w, c in most_common))

    return total, freq, most_common[0][0], most_common[0][1]

decision_word_count, decision_word_freq, decision_most_common, decision_highest_freq = analyze_text("Decision Tree", decision_tree_text)
forest_word_count, forest_word_freq, forest_most_common, forest_highest_freq = analyze_text("Random Forest", random_forest_text)
boosting_word_count, boosting_word_freq, boosting_most_common, boosting_highest_freq = analyze_text("Boosting Trees", boosting_trees_text)
```
