Here we get the first few rows of our data, aswell as the total number of inversion periods. We see that an inversion is a uncommon event. 

```python
df = pd.read_csv("D:/Users/nickt/ECO482/Research Project Files/Cleaning/FRED_MD_Cleaned.csv")

df_1 = pd.read_csv("D:/Users/nickt/ECO482/Research Project Files/Cleaning/FRED_MD_Cleaned_NonTransformed.csv")

display(df.head())

df['Inversion'].value_counts()
```

Below we plot the Slope variable, to see the yield curve inversions over time. All points below the black line are times when the yield curve was inverted. 

```python
plt = px.line(df, x="sasdate", y="Slope", title="Slope of the Yield Curve Over Time")

plt.add_hline(y=0)

plt.show()
```

We can display some summary statistics of our cleaned data (Pre Transformation). Keep in mind, these are the transformed values of our data.

```python
cols = ['GS10', 'TB3MS', 'Slope', 'UNRATE', 'S&P 500']

summary_stats = df_1[cols].agg(['mean', 'std', 'min', 'max']).transpose()

summary_stats.columns = ['Mean', 'Std Dev', 'Min', 'Max']

print("="*50)
print("DATASET SUMMARY STATISTICS")
print("="*50)
print(summary_stats.round(2))

print(len(df_1.columns))

display(df_1[cols].corr())
```
